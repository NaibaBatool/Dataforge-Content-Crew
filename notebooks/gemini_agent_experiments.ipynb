# DataForge Content Crew — Gemini API & Structured Output Experiments

This Jupyter notebook demonstrates how to connect to the Google Gemini API using the official `google.genai` SDK and perform structured JSON output experiments for the Content Crew pipeline.

### How to Obtain a Gemini API Key:
1. Navigate to [Google AI Studio](https://aistudio.google.com/).
2. Sign in with your Google account.
3. Click **Get API key** and create a key in a Google Cloud project.
4. Set `GEMINI_API_KEY` in your environment or `.env` file.

```python
import os
import json
from google import genai
from google.genai import types

# 1. Read API credentials
api_key = os.environ.get("GEMINI_API_KEY", "").strip()
model_name = os.environ.get("GEMINI_MODEL", "gemini-2.5-flash")

print(f"Target Model: {model_name}")
print(f"API Key Configured: {bool(api_key and api_key != 'your_gemini_api_key_here')}")
```

## 2. Basic Text Generation Experiment
Send a concise editorial prompt to verify connectivity and observe latency.

```python
if api_key and api_key != "your_gemini_api_key_here":
    client = genai.Client(api_key=api_key)
    response = client.models.generate_content(
        model=model_name,
        contents="Explain why multi-cloud Kubernetes deployments require unified GitOps in 3 sentences."
    )
    print("--- Gemini Response ---")
    print(response.text)
else:
    print("GEMINI_API_KEY not set. Running in offline demonstration mode.")
```

## 3. Structured JSON Output Experiment
Enforce JSON output using `response_mime_type="application/json"` to generate a structured competitor and outline matrix.

```python
if api_key and api_key != "your_gemini_api_key_here":
    prompt = """
    Generate an outline for a B2B article titled 'Hybrid Cloud Observability'.
    Return JSON with: title (str), target_audience (str), key_takeaways (array of str), outline_sections (array of str).
    """
    config = types.GenerateContentConfig(
        temperature=0.2,
        response_mime_type="application/json"
    )
    response = client.models.generate_content(
        model=model_name,
        contents=prompt,
        config=config
    )
    structured_data = json.loads(response.text)
    print(json.dumps(structured_data, indent=2))
else:
    print("Structured output experiment simulated offline.")
    mock_output = {
        "title": "Hybrid Cloud Observability: Unified Telemetry Guide",
        "target_audience": "VP Engineering and DevOps Architects",
        "key_takeaways": [
            "Eliminate telemetry silos across AWS and on-prem clusters.",
            "Correlate distributed traces with structured error logs."
        ],
        "outline_sections": ["Introduction", "The Multi-Cloud Blindspot", "Unified Architecture", "Conclusion"]
    }
    print(json.dumps(mock_output, indent=2))
```
